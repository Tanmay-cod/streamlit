```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name":["Amit", "Bharat", "Chain", "Dean", "Emma"],
    "maths":[90, 80, 70, 60, 50],
    "science":[85, 75, 65, 55, 45],
    "english":[95, 85, 75, 65, 55]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
marks = df[["maths", "science", "english"]].values
average_marks = np.mean(marks, axis=1)
df["average"]=average_marks
print(df)
```

```python
#Let's assign the grade
def grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 75:
        return 'B'
    elif avg >= 60:
        return 'C'
    else:
        return 'D'


df['grade'] = df["average"].apply(grade)
df
```

```python
#Lets make the graph
plt.figure()
plt.bar(df["Name"], df["average"])
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```

```python
# avg marks per subject
df[[ "maths", "science", "english"]].mean().plot(kind='bar')
plt.title("Average Marks per Subject")
plt.xlabel("Subjects")
plt.ylabel(" Marks")
plt.show()
```
