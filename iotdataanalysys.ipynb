```python
# import lib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load iot data 
df = pd.read_csv('iot_sensor_data.csv')
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
print("iot_sensor_data\n",df)
```

```python
#staticstical analysys
temp = df['Temperature'].values
mean_temp = np.mean(temp)
std_temp = np.std(temp)
print ("Mean Temperature:", mean_temp)
print ("Standard Deviation of Temperature:", std_temp)
```

```python
# Line graph of Temperature over time
plt.figure(figsize=(12, 6))
plt.plot(df["Timestamp"], df["Temperature"], marker="o", color="blue")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Trend Over Time")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
